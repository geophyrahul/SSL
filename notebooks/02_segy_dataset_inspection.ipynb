```python
import segyio
import numpy as np

fname = "../data/Teapot_dome.sgy"

with segyio.open(fname, "r", ignore_geometry=True) as f:

    ilines = np.array([
        f.header[i][segyio.TraceField.INLINE_3D]
        for i in range(f.tracecount)
    ])

    xlines = np.array([
        f.header[i][segyio.TraceField.CROSSLINE_3D]
        for i in range(f.tracecount)
    ])

print("Unique inlines:", len(np.unique(ilines)))
print("Unique xlines:", len(np.unique(xlines)))

print("Inline range:", ilines.min(), ilines.max())
print("Xline range:", xlines.min(), xlines.max())
```

```python
import segyio
import numpy as np

fname = "../data/Teapot_dome.sgy"

with segyio.open(fname, "r", ignore_geometry=True) as f:

    cdp = np.array([
        f.header[i][segyio.TraceField.CDP]
        for i in range(f.tracecount)
    ])

print("Unique CDPs:", len(np.unique(cdp)))
print("CDP min:", cdp.min())
print("CDP max:", cdp.max())
```

```python
import segyio
import numpy as np

fname = "../data/Teapot_dome.sgy"

with segyio.open(fname, "r", ignore_geometry=True) as f:

    xs = np.array([
        f.header[i][segyio.TraceField.SourceX]
        for i in range(f.tracecount)
    ])

    ys = np.array([
        f.header[i][segyio.TraceField.SourceY]
        for i in range(f.tracecount)
    ])

print("Unique X:", len(np.unique(xs)))
print("Unique Y:", len(np.unique(ys)))

print(xs.min(), xs.max())
print(ys.min(), ys.max())
```

```python
import segyio
import numpy as np

fname = "../data/Teapot_dome.sgy"

with segyio.open(fname, "r", ignore_geometry=True) as f:

    traces = segyio.tools.collect(f.trace[:])

print("Trace matrix:", traces.shape)
```

```python
cube = traces.reshape(345, 188, 1501)

print(cube.shape)
```

```python
np.save("../data/teapot_dome.npy", cube)
```

```python
import matplotlib.pyplot as plt

cube = np.load("../data/teapot_dome.npy")

plt.figure(figsize=(10,6))
plt.imshow(
    cube[100].T,
    cmap="gray",
    aspect="auto"
)
plt.title("Inline 100")
plt.show()
```
